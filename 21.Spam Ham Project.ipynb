```python
def vocab(corpus):
    vocab = []
    for document in corpus:
        for word in document.split():
            if word not in vocab:
                vocab.append(word)
    vocab.sort()
    v = {w:v for v, w in enumerate(vocab)}
    return v
```

```python
def bag_of_word(corpus, vocabulary):
    vectors = []
    for document in corpus:
        vector = [0] * len(vocabulary)
        for word in document.split():
            if word in vocabulary:
                i = vocabulary[word]
                vector[i] += 1
        vectors.append(vector)
    return vectors
```

```python
from sklearn.base import BaseEstimator, TransformerMixin
```

```python
class Vectorization(BaseEstimator, TransformerMixin):
    def __init__(self):
        self.vocab_ = None
        self.vocab_size_ = None

    def fit(self, corpus):
        vocab = []
        for document in corpus:
            for word in document.split():
                if word not in vocab:
                    vocab.append(word)
        vocab.sort()
        self.vocab_ = {w:v for v, w in enumerate(vocab)}
        self.vocab_size_ = len(self.vocab_)

    def transform(self, corpus):
        vectors = []
        for document in corpus:
            vector = [0] * self.vocab_size_
            for word in document.split():
                if word in self.vocab_:
                    i = self.vocab_[word]
                    vector[i] += 1
            vectors.append(vector)
        return vectors
```
